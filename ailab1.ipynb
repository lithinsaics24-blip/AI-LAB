```python
# Tic-Tac-Toe Game

board = [' ' for i in range(9)]

def display_board():
    print()
    print(board[0], '|', board[1], '|', board[2])
    print('--+---+--')
    print(board[3], '|', board[4], '|', board[5])
    print('--+---+--')
    print(board[6], '|', board[7], '|', board[8])
    print()

def check_winner(player):
    # Rows
    if board[0] == board[1] == board[2] == player:
        return True
    if board[3] == board[4] == board[5] == player:
        return True
    if board[6] == board[7] == board[8] == player:
        return True

    # Columns
    if board[0] == board[3] == board[6] == player:
        return True
    if board[1] == board[4] == board[7] == player:
        return True
    if board[2] == board[5] == board[8] == player:
        return True

    # Diagonals
    if board[0] == board[4] == board[8] == player:
        return True
    if board[2] == board[4] == board[6] == player:
        return True

    return False


player = 'X'

for turn in range(9):

    display_board()

    position = int(input("Player " + player + ", enter position (1-9): ")) - 1

    if board[position] == ' ':
        board[position] = player
    else:
        print("Position already occupied!")
        continue

    if check_winner(player):
        display_board()
        print("Player", player, "wins!")
        break

    if turn == 8:
        display_board()
        print("Game is a draw!")

    if player == 'X':
        player = 'O'
    else:
        player = 'X'
```

```python
# Vacuum Cleaner - Goal Based Agent

room = {
    'A': 'Dirty',
    'B': 'Dirty'
}

location = 'A'

goal = 'All rooms should be Clean'

while True:

    print("\nCurrent Location:", location)
    print("Room A:", room['A'])
    print("Room B:", room['B'])

    # Check whether the goal is achieved
    if room['A'] == 'Clean' and room['B'] == 'Clean':
        print("\nGoal:", goal)
        print("Goal achieved!")
        break

    if room[location] == 'Dirty':
        print("Room is Dirty")
        print("Action: Suck")
        room[location] = 'Clean'

    else:
        print("Room is Clean")

        if location == 'A':
            location = 'B'
            print("Action: Move to B")
        else:
            location = 'A'
            print("Action: Move to A")
```
