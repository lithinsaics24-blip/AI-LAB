```python
import random

def get_heuristic(state):

    n = len(state)
    attacking_pairs = 0
    for i in range(n):
        for j in range(i + 1, n):

            if state[i] == state[j]:
                attacking_pairs += 1

            elif abs(state[i] - state[j]) == abs(i - j):
                attacking_pairs += 1
    return attacking_pairs

def get_best_neighbor(state):

    n = len(state)
    best_state = list(state)
    best_h = get_heuristic(state)


    for col in range(n):
        original_row = state[col]
        for row in range(n):
            if row != original_row:
                neighbor = list(state)
                neighbor[col] = row
                neighbor_h = get_heuristic(neighbor)

                if neighbor_h < best_h:
                    best_h = neighbor_h
                    best_state = neighbor

    return best_state, best_h

def hill_climbing_n_queens(n=8, max_restarts=100):

    for restart in range(max_restarts):

        current_state = [random.randint(0, n - 1) for _ in range(n)]
        current_h = get_heuristic(current_state)

        while True:
            neighbor, neighbor_h = get_best_neighbor(current_state)

            if neighbor_h == 0:
                print(f"Solution found after {restart} restart(s)!")
                return neighbor


            if neighbor_h >= current_h:
                break

            current_state = neighbor
            current_h = neighbor_h

    print("Failed to find a solution within the maximum restarts limit.")
    return None

def print_board(state):

    if state is None:
        return
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)


n = 8
solution = hill_climbing_n_queens(n)
if solution:
    print(f"\nState representation (row index per column): {solution}")
    print("\nVisualized Board:")
    print_board(solution)
```
