```python
def is_variable(x):
    """Checks if a term is a variable (represented as a string starting with a lowercase letter)."""
    return isinstance(x, str) and len(x) > 0 and x[0].islower()

def is_constant(x):
    """Checks if a term is a constant (represented as a string starting with an uppercase letter)."""
    return isinstance(x, str) and len(x) > 0 and x[0].isupper()

def is_compound(x):
    """Checks if a term is compound (represented as a tuple, e.g., ('f', 'x', 'y'))."""
    return isinstance(x, tuple) and len(x) > 0

def occur_check(var, term, subst):
    """Returns True if variable var occurs anywhere inside term under substitution subst."""
    if var == term:
        return True
    elif is_variable(term):
        if term in subst:
            return occur_check(var, subst[term], subst)
        return False
    elif is_compound(term):
        # Check function name and all arguments
        return any(occur_check(var, arg, subst) for arg in term[1:])
    return False

def unify(x, y, subst=None):
    """Recursively unifies expressions x and y with an optional starting substitution.
    Returns the substitution dict if successful, or None if unification fails.
    """
    if subst is None:
        subst = {}

    # If substitution is already failure
    if subst is False:
        return False

    # Step 1: Identical terms
    if x == y:
        return subst

    # Step 2: If x is a variable
    elif is_variable(x):
        return unify_var(x, y, subst)

    # Step 3: If y is a variable
    elif is_variable(y):
        return unify_var(y, x, subst)

    # Step 4: Compound terms (Functions / Predicates)
    elif is_compound(x) and is_compound(y):
        # Must have the same function name/predicate and same number of arguments
        if len(x) != len(y) or x[0] != y[0]:
            return False

        # Unify arguments sequentially
        for arg_x, arg_y in zip(x[1:], y[1:]):
            subst = unify(arg_x, arg_y, subst)
            if subst is False:
                return False
        return subst

    # Step 5: Constants or mismatch
    else:
        return False

def unify_var(var, val, subst):
    """Helper to unify a variable var with a value val under substitution subst."""
    if var in subst:
        return unify(subst[var], val, subst)
    elif is_variable(val) and val in subst:
        return unify(var, subst[val], subst)
    elif occur_check(var, val, subst):
        print(f"Occur check failed: {var} is inside {val}")
        return False
    else:
        # Add new mapping
        new_subst = subst.copy()
        new_subst[var] = val
        return new_subst
```

```python
import ast

def is_variable(x):
    return isinstance(x, str) and len(x) > 0 and x[0].islower()

def is_constant(x):
    return isinstance(x, str) and len(x) > 0 and x[0].isupper()

def is_compound(x):
    return isinstance(x, tuple) and len(x) > 0

def occur_check(var, term, subst):
    if var == term:
        return True
    elif is_variable(term):
        if term in subst:
            return occur_check(var, subst[term], subst)
        return False
    elif is_compound(term):
        return any(occur_check(var, arg, subst) for arg in term[1:])
    return False

def unify(x, y, subst=None):
    if subst is None:
        subst = {}
    if subst is False:
        return False
    if x == y:
        return subst
    elif is_variable(x):
        return unify_var(x, y, subst)
    elif is_variable(y):
        return unify_var(y, x, subst)
    elif is_compound(x) and is_compound(y):
        if len(x) != len(y) or x[0] != y[0]:
            return False
        for arg_x, arg_y in zip(x[1:], y[1:]):
            subst = unify(arg_x, arg_y, subst)
            if subst is False:
                return False
        return subst
    else:
        return False

def unify_var(var, val, subst):
    if var in subst:
        return unify(subst[var], val, subst)
    elif is_variable(val) and val in subst:
        return unify(var, subst[val], subst)
    elif occur_check(var, val, subst):
        print(f"Occur check failed: {var} is inside {val}")
        return False
    else:
        new_subst = subst.copy()
        new_subst[var] = val
        return new_subst

def parse_term(input_str):
    try:
        return ast.literal_eval(input_str.strip())
    except Exception:
        return input_str.strip()

print("Enter terms to unify.")
print("Examples:")
print("  Term 1: ('f', 'x')")
print("  Term 2: ('f', 'A')")
print("-" * 30)

try:
    term1_raw = input("Enter Term 1: ") or "('f', 'x')"
    term2_raw = input("Enter Term 2: ") or "('f', 'A')"

    term1 = parse_term(term1_raw)
    term2 = parse_term(term2_raw)

    print(f"\nParsed Term 1: {term1}")
    print(f"Parsed Term 2: {term2}")

    result = unify(term1, term2)
    if result is False:
        print("\nResult: Unification Failed!")
    else:
        print(f"\nResult: Unification Successful!")
        print(f"Most General Unifier (MGU): {result}")
except Exception as e:
    print(f"An error occurred: {e}")







```
